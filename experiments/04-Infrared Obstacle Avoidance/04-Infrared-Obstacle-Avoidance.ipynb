```python
import RPi.GPIO as GPIO
import time

SENSOR_LEFT = 12
SENSOR_RIGHT = 16

GPIO.setwarnings(False)
GPIO.setmode(GPIO.BCM)

GPIO.setup(SENSOR_LEFT, GPIO.IN)
GPIO.setup(SENSOR_RIGHT, GPIO.IN)

print("Infrared sensors initialized.")
```

```python
start_time = time.monotonic()

try:
    while time.monotonic() - start_time < 5:

        left = GPIO.input(SENSOR_LEFT)
        right = GPIO.input(SENSOR_RIGHT)

        print(f"Left: {left} | Right: {right}")

        time.sleep(0.2)

finally:
    GPIO.cleanup()
    print("5-second sensor test finished.")
```

```python
import RPi.GPIO as GPIO
import time

LEFT_SENSOR = 12
RIGHT_SENSOR = 16

GPIO.setwarnings(False)
GPIO.setmode(GPIO.BCM)

GPIO.setup(LEFT_SENSOR, GPIO.IN)
GPIO.setup(RIGHT_SENSOR, GPIO.IN)

try:
    for i in range(40):
        left = GPIO.input(LEFT_SENSOR)
        right = GPIO.input(RIGHT_SENSOR)

        print(f"Left: {left} | Right: {right}")
        time.sleep(0.25)

finally:
    GPIO.cleanup()
    print("Sensor test completed.")
```

```python
def decide_action(left, right):
    if left == 1 and right == 1:
        return "FORWARD"

    elif left == 0 and right == 1:
        return "TURN_RIGHT"

    elif left == 1 and right == 0:
        return "TURN_LEFT"

    else:
        return "STOP"
```

```python
test_cases = [
    (1, 1),
    (0, 1),
    (1, 0),
    (0, 0)
]

for left, right in test_cases:
    action = decide_action(left, right)
    print(f"Left={left}, Right={right} -> {action}")
```

```python
def execute_action(robot, action, speed=25):
    if action == "FORWARD":
        robot.t_up(speed, 0.1)

    elif action == "TURN_LEFT":
        robot.turnLeft(speed, 0.1)

    elif action == "TURN_RIGHT":
        robot.turnRight(speed, 0.1)

    else:
        robot.t_stop(0.1)
```

```python
def execute_action(robot, action, speed=25, duration=0.3):
    try:
        if action == "FORWARD":
            robot.t_up(speed, duration)

        elif action == "TURN_LEFT":
            robot.turnLeft(speed, duration)

        elif action == "TURN_RIGHT":
            robot.turnRight(speed, duration)

        elif action == "STOP":
            robot.t_stop(0)

        else:
            raise ValueError(f"Unknown action: {action}")

    finally:
        robot.t_stop(0)
```

```python
import RPi.GPIO as GPIO
import time

LEFT_SENSOR = 12
RIGHT_SENSOR = 16

GPIO.setwarnings(False)
GPIO.setmode(GPIO.BCM)

GPIO.setup(LEFT_SENSOR, GPIO.IN)
GPIO.setup(RIGHT_SENSOR, GPIO.IN)

# Initial test settings
SPEED = 25
DURATION = 10
ACTION_TIME = 0.2

start_time = time.monotonic()

try:
    robot.t_stop(0)

    print("Autonomous obstacle avoidance started!")

    while time.monotonic() - start_time < DURATION:

        # Read infrared sensors
        left = GPIO.input(LEFT_SENSOR)
        right = GPIO.input(RIGHT_SENSOR)

        # Make decision
        action = decide_action(left, right)

        # Display sensor data and decision
        print(
            f"Left: {left} | "
            f"Right: {right} | "
            f"Action: {action}"
        )

        # Execute motor command
        execute_action(
            robot,
            action,
            speed=SPEED,
            duration=ACTION_TIME
        )

        time.sleep(0.05)

except KeyboardInterrupt:
    print("Interrupted by user.")

finally:
    robot.t_stop(0)
    GPIO.cleanup()
    print("Robot stopped. Test completed.")
```
