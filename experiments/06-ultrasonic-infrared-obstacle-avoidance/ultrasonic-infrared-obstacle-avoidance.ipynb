```python
import time
import RPi.GPIO as GPIO

TRIG = 20
ECHO = 21

def measure_distance(timeout=0.03):
    GPIO.output(TRIG, GPIO.LOW)
    time.sleep(0.000002)

    # Send a 10-microsecond trigger pulse
    GPIO.output(TRIG, GPIO.HIGH)
    time.sleep(0.00001)
    GPIO.output(TRIG, GPIO.LOW)

    # Wait for ECHO to go HIGH
    deadline = time.monotonic() + timeout

    while GPIO.input(ECHO) == GPIO.LOW:
        if time.monotonic() > deadline:
            return None

    start = time.monotonic()

    # Wait for ECHO to go LOW
    deadline = start + timeout

    while GPIO.input(ECHO) == GPIO.HIGH:
        if time.monotonic() > deadline:
            return None

    end = time.monotonic()

    distance = (end - start) * 34300 / 2

    return round(distance, 1)
```

```python
import RPi.GPIO as GPIO
import time
from LOBOROBOT import LOBOROBOT

LEFT = 12
RIGHT = 16
TRIG = 20
ECHO = 21

SPEED = 25
DURATION = 0.2
TEST_CYCLES = 20

robot = None

try:
    # Initialize robot
    robot = LOBOROBOT()
    robot.t_stop(0)

    # Initialize sensors
    GPIO.setwarnings(False)
    GPIO.setmode(GPIO.BCM)

    GPIO.setup(LEFT, GPIO.IN)
    GPIO.setup(RIGHT, GPIO.IN)
    GPIO.setup(TRIG, GPIO.OUT)
    GPIO.setup(ECHO, GPIO.IN)
    GPIO.output(TRIG, GPIO.LOW)

    print("Experiment 06 started!")

    for i in range(TEST_CYCLES):
        left = GPIO.input(LEFT)
        right = GPIO.input(RIGHT)
        distance = measure_distance()

        action = combined_decision(left, right, distance)

        print(
            f"Test {i+1:02d} | "
            f"L={left} R={right} | "
            f"D={distance} cm | "
            f"Action={action}"
        )

        # Stop before every new action
        robot.t_stop(0)

        try:
            if action == "FORWARD":
                robot.t_up(SPEED, DURATION)

            elif action == "TURN_LEFT":
                robot.turnLeft(SPEED, DURATION)

            elif action == "TURN_RIGHT":
                robot.turnRight(SPEED, DURATION)

            elif action == "STOP":
                time.sleep(DURATION)

        finally:
            robot.t_stop(0)

        time.sleep(0.1)

except KeyboardInterrupt:
    print("Experiment interrupted.")

finally:
    try:
        if robot is not None:
            robot.t_stop(0)
    finally:
        GPIO.cleanup()

    print("Experiment 06 test ended.")
```

```python
import time
import RPi.GPIO as GPIO

TRIG = 20
ECHO = 21

def measure_distance_v2(timeout=0.03):
    GPIO.output(TRIG, GPIO.LOW)
    time.sleep(0.000005)

    GPIO.output(TRIG, GPIO.HIGH)
    time.sleep(0.00001)
    GPIO.output(TRIG, GPIO.LOW)

    deadline = time.monotonic() + timeout

    # Wait for rising edge
    while GPIO.input(ECHO) == GPIO.LOW:
        if time.monotonic() >= deadline:
            return None

    start = time.monotonic()
    deadline = start + timeout

    # Wait for falling edge
    while GPIO.input(ECHO) == GPIO.HIGH:
        if time.monotonic() >= deadline:
            return None

    end = time.monotonic()

    distance = (end - start) * 34300 / 2

    # Reject invalid readings
    if not (2 <= distance <= 400):
        return None

    return round(distance, 1)

print("measure_distance_v2 defined successfully!")
```

```python
import RPi.GPIO as GPIO
import time
from LOBOROBOT import LOBOROBOT

LEFT = 12
RIGHT = 16
TRIG = 20
ECHO = 21

THRESHOLD = 40
SPEED = 25
MOVE_DURATION = 0.2
TEST_CYCLES = 30

def combined_decision(left, right, distance):
    if distance is None or not (2 <= distance <= 400):
        return "STOP"

    if distance < THRESHOLD:
        return "STOP"

    if left == 0 and right == 0:
        return "STOP"

    if left == 0:
        return "TURN_RIGHT"

    if right == 0:
        return "TURN_LEFT"

    return "FORWARD"


def execute_action(robot, action):
    robot.t_stop(0)

    try:
        if action == "FORWARD":
            robot.t_up(SPEED, MOVE_DURATION)

        elif action == "TURN_LEFT":
            robot.turnLeft(SPEED, MOVE_DURATION)

        elif action == "TURN_RIGHT":
            robot.turnRight(SPEED, MOVE_DURATION)

        else:
            time.sleep(MOVE_DURATION)

    finally:
        robot.t_stop(0)


robot = None
completed = False

try:
    # Initialize motor controller
    robot = LOBOROBOT()
    robot.t_stop(0)

    # Initialize GPIO
    GPIO.setwarnings(False)
    GPIO.setmode(GPIO.BCM)

    GPIO.setup(LEFT, GPIO.IN)
    GPIO.setup(RIGHT, GPIO.IN)
    GPIO.setup(TRIG, GPIO.OUT)
    GPIO.setup(ECHO, GPIO.IN)
    GPIO.output(TRIG, GPIO.LOW)

    print("Experiment 06 motor test started!")

    for i in range(TEST_CYCLES):

        left = GPIO.input(LEFT)
        right = GPIO.input(RIGHT)
        distance = measure_distance_v2()

        action = combined_decision(left, right, distance)

        print(
            f"Test {i+1:02d} | "
            f"L={left} R={right} | "
            f"D={distance} cm | "
            f"Action={action}"
        )

        execute_action(robot, action)

        time.sleep(0.1)

    completed = True

except KeyboardInterrupt:
    print("Experiment interrupted.")

finally:
    try:
        if robot is not None:
            robot.t_stop(0)
    finally:
        GPIO.cleanup()

    print(
        "Experiment 06 motor test completed."
        if completed
        else "Experiment 06 motor test stopped."
    )
```

```python
from LOBOROBOT import LOBOROBOT
import RPi.GPIO as GPIO
import time

robot = None

test_cases = [
    ("FORWARD", 1, 1, 60),
    ("TURN_RIGHT", 0, 1, 60),
    ("TURN_LEFT", 1, 0, 60),
    ("STOP_BOTH_IR", 0, 0, 60),
    ("STOP_ULTRASONIC", 1, 1, 20),
    ("STOP_TIMEOUT", 1, 1, None),
]

try:
    robot = LOBOROBOT()
    robot.t_stop(0)

    for name, left, right, distance in test_cases:
        action = combined_decision(left, right, distance)

        print(f"Testing {name}: {action}")

        execute_action(robot, action)

        print("Motor stopped.")
        time.sleep(1)

finally:
    try:
        if robot is not None:
            robot.t_stop(0)
    finally:
        GPIO.cleanup()

print("Motor direction test completed.")
```
