
***1. 做“滴滴滴”提示音：学习如何控制“响/停”的节奏。***

```python
import RPi.GPIO as GPIO
import time

BUZZER = 11

GPIO.setwarnings(False)
GPIO.setmode(GPIO.BOARD)
GPIO.setup(BUZZER, GPIO.OUT)

buzzer = GPIO.PWM(BUZZER, 440)

# 滴滴滴
for i in range(3):
    buzzer.start(50)
    time.sleep(0.1)

    buzzer.stop()
    time.sleep(0.1)

GPIO.cleanup()
```

***2. 封装成函数：从一坨实验代码变成 beep()、play_note()、play_song()。***

```python
import RPi.GPIO as GPIO
import time

BUZZER = 11

GPIO.setwarnings(False)
GPIO.setmode(GPIO.BOARD)
GPIO.setup(BUZZER, GPIO.OUT)

buzzer = GPIO.PWM(BUZZER, 440)
buzzer.start(0)  # PWM保持运行，但一开始不发声


def beep(times=1, frequency=440, duration=0.1, interval=0.1):
    """
    Make the buzzer beep.

    times:      number of beeps
    frequency:  frequency in Hz
    duration:   duration of each beep in seconds
    interval:   silence between beeps in seconds
    """

    buzzer.ChangeFrequency(frequency)

    for _ in range(times):
        # Sound ON
        buzzer.ChangeDutyCycle(50)
        time.sleep(duration)

        # Sound OFF
        buzzer.ChangeDutyCycle(0)
        time.sleep(interval)
```

```python
beep(3)
```

***3.做小车提示音系统：比如开机 startup()、警告 warning()、成功 success()。***

```python
def startup_sound():
    beep(times=2, frequency=600, duration=0.1, interval=0.1)


def warning_sound():
    beep(times=5, frequency=900, duration=0.05, interval=0.05)


def success_sound():
    beep(times=2, frequency=1200, duration=0.15, interval=0.1)
```
