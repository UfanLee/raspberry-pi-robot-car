```python
print("Testing FORWARD-LEFT...")

robot.MotorStop(0)
robot.MotorRun(1, 'forward', 30)
robot.MotorRun(2, 'forward', 30)
robot.MotorStop(3)

import time
time.sleep(1)

robot.t_stop(1)

print("Forward-left test finished.")
```

```python
import time

for motor in range(4):
    print(f"Testing Motor {motor}")

    robot.MotorRun(motor, 'forward', 25)
    time.sleep(0.5)

    robot.MotorStop(motor)
    time.sleep(1)
```

```python
print("Testing FORWARD-RIGHT...")

robot.MotorRun(0, 'forward', 30)
robot.MotorStop(1)
robot.MotorStop(2)
robot.MotorRun(3, 'forward', 30)

import time
time.sleep(1)

robot.t_stop(1)

print("Forward-right test finished.")
```

```python
M0 左前 → Backward
M1 右前 → Stop
M2 左后 → Stop
M3 右后 → Backward
```
